```python
import pandas as pd
import numpy as np

# Carregando tabela
agencia = pd.read_csv('/agencias.csv')
clientes = pd.read_csv('/clientes.csv')
contas = pd.read_csv('/contas.csv')
transacoes = pd.read_csv('/transacoes.csv')

# Merge
df_banco = pd.merge(contas, clientes, on='cod_cliente', how='left')
df_final = pd.merge(transacoes, df_banco, on='num_conta', how='inner')
df_final.head(10)
```

```python
# Definir o intervalo de datas baseado nos dados do BanVic
data_min = df_final['data_transacao'].min()
data_max = df_final['data_transacao'].max()

# Criar a dim_dates
dim_dates = pd.DataFrame({'data': pd.date_range(start=data_min, end=data_max)})

# Adicionar atributos exigidos pelo desafio
dim_dates['ano'] = dim_dates['data'].dt.year
dim_dates['mes'] = dim_dates['data'].dt.month
dim_dates['dia_da_semana'] = dim_dates['data'].dt.day_name() # Para responder à média de transações
dim_dates['nome_mes'] = dim_dates['data'].dt.month_name()

# Validar a afirmação do analista sobre meses pares vs ímpares
dim_dates['mes_par'] = dim_dates['mes'] % 2 == 0

print(dim_dates.head())
```

```python
# Substitua 'valor_transacao' pelo nome real da coluna de valor no seu dataset
volume_por_conta = transacoes.groupby('num_conta')['valor_transacao'].sum().reset_index()

# Unir (Merge) o volume transacionado com a tabela de contas (que já está com a data limpa)
df_analise = pd.merge(contas, volume_por_conta, on='num_conta', how='left')

# Preencher com 0 as contas que nunca transacionaram
df_analise['valor_transacao'] = df_analise['valor_transacao'].fillna(0)

# Criar uma métrica de "Anos de Conta" para facilitar a visualização
# Precisamos converter de volta para datetime para calcular a diferença
hoje = pd.to_datetime('today')
df_analise['data_abertura_dt'] = pd.to_datetime(df_analise['data_abertura'])
df_analise['anos_de_conta'] = (hoje - df_analise['data_abertura_dt']).dt.days // 365

# Agrupar por ano de abertura para ver a média de transação por "safra" de clientes
ranking_antiguidade = df_analise.groupby('anos_de_conta')['valor_transacao'].mean().sort_index(ascending=False)

print("Média de Volume Transacionado por Tempo de Casa (Anos):")
print(ranking_antiguidade)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Criar o ano de abertura
df_final['ano_abertura'] = pd.to_datetime(df_final['data_abertura']).dt.year

# Agrupar por ano e calcular volume total e ticket médio
analise_safra = df_final.groupby('ano_abertura').agg({
    'valor_transacao': ['sum', 'mean', 'count'],
    'nome_completo': 'count'
}).reset_index()

analise_safra.columns = ['Ano', 'Volume_Total', 'Ticket_Medio', 'Num_Transacoes', 'Qtd_Clientes']

# Visualização para o Relatório
plt.figure(figsize=(12, 6))
sns.barplot(data=analise_safra, x='Ano', y='Volume_Total', palette='winter')
plt.title('Volume Financeiro por Safra de Cliente (Ano de Abertura)')
plt.show()
```

```python
import pandas as pd

# Garantir que a data está no formato correto e criar a coluna de Ano
transacoes['data_transacao'] = pd.to_datetime(transacoes['data_transacao'], errors='coerce')
transacoes['ano_transacao'] = transacoes['data_transacao'].dt.year.fillna(0).astype(int)

# Vamos ver o Valor Total e a Quantidade de Transações por Ano
planilha_resumo = transacoes.groupby('ano_transacao').agg({
    'valor_transacao': ['sum', 'count', 'mean']
}).reset_index()

# Renomear as colunas para ficarem profissionais
planilha_resumo.columns = ['Ano', 'Volume_Total', 'Qtd_Transacoes', 'Ticket_Medio']

# 3. Ordenar do ano mais recente para o mais antigo
planilha_resumo = planilha_resumo.sort_values(by='Ano', ascending=False)

print("--- RESUMO DE TRANSAÇÕES POR ANO ---")
print(planilha_resumo)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Criar o gráfico de barras (Volume Total)
ax = sns.barplot(data=planilha_resumo, x='Ano', y='Volume_Total', palette='Blues_d')

# Adicionar uma linha para o Ticket Médio (Eixo secundário)
ax2 = ax.twinx()
sns.lineplot(data=planilha_resumo, x=range(len(planilha_resumo)), y='Ticket_Medio',
             color='orange', marker='o', ax=ax2, label='Ticket Médio')

# Customização de títulos e legendas
ax.set_title('Evolução Financeira BanVic: Volume Total vs. Ticket Médio', fontsize=16, fontweight='bold')
ax.set_xlabel('Ano de Transação', fontsize=12)
ax.set_ylabel('Volume Total (R$)', fontsize=12)
ax2.set_ylabel('Ticket Médio (R$)', fontsize=10)

# Adicionar rótulos de valor nas barras para facilitar a leitura
for p in ax.patches:
    ax.annotate(f'R${p.get_height():,.0f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 9), textcoords='offset points', fontsize=10)

plt.tight_layout()
plt.show()
```
