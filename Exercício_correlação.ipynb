```python
import pandas as pd
from sklearn import datasets
file_path = '/mnt/data/eWB4AbjCQwuYUbQJY8He_iris.csv'
iris_df = pd.read_csv(file_path)

iris_df.head(), iris_df.describe(), iris_df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Boxplots of Iris Features by Species')

sns.boxplot(x='Species', y='Sepal_Length', data=iris_df, ax=axs[0, 0])
sns.boxplot(x='Species', y='Sepal_Width', data=iris_df, ax=axs[0, 1])
sns.boxplot(x='Species', y='Petal_Length', data=iris_df, ax=axs[1, 0])
sns.boxplot(x='Species', y='Petal_Width', data=iris_df, ax=axs[1, 1])

axs[0, 0].set_title('Sepal Length')
axs[0, 1].set_title('Sepal Width')
axs[1, 0].set_title('Petal Length')
axs[1, 1].set_title('Petal Width')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()

fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Histograms of Iris Features by Species')

sns.histplot(data=iris_df, x='Sepal_Length', hue='Species', kde=True, ax=axs[0, 0])
sns.histplot(data=iris_df, x='Sepal_Width', hue='Species', kde=True, ax=axs[0, 1])
sns.histplot(data=iris_df, x='Petal_Length', hue='Species', kde=True, ax=axs[1, 0])
sns.histplot(data=iris_df, x='Petal_Width', hue='Species', kde=True, ax=axs[1, 1])

axs[0, 0].set_title('Sepal Length')
axs[0, 1].set_title('Sepal Width')
axs[1, 0].set_title('Petal Length')
axs[1, 1].set_title('Petal Width')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

X = iris_df.drop(columns='Species')
y = iris_df['Species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

log_reg_model = LogisticRegression(max_iter=200, random_state=42)
log_reg_model.fit(X_train, y_train)

y_pred = log_reg_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')

accuracy, precision, recall
```

```python
conf_matrix = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=log_reg_model.classes_, yticklabels=log_reg_model.classes_)
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix for Logistic Regression Model")
plt.show()
```
